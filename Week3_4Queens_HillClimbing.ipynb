```python
row = []
print("Enter 4 position of queen in rows (0-3)")
for i in range(0, 4):
  row.append(int(input(f"{i} column : ")))
print("Current state:", row)

def print_board(state):
  for r in range(4):
    line = ""
    for c in range(4):
      if state[c] == r:
        line += " Q "
      else:
        line += " . "
    print(line)
  print()

print("\nMapping for your input:")
print_board(row)

conflicts = 0
for i in range(4):
  for j in range(i + 1, 4):
    if row[i] == row[j]:
      conflicts += 1
    elif abs(row[i] - row[j]) == abs(i - j):
      conflicts += 1

print(f"Total conflicts: {conflicts}")
if conflicts == 0:
  print("Success!")
else:
  print("Correct solution board mapping:")
  print_board([1, 3, 0, 2])
```
