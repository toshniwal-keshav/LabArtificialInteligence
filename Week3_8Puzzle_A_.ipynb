```python
import heapq

def manhattan(state, goal):
    dist = 0
    for i, val in enumerate(state):
        if val != 0:
            r, c = divmod(i, 3)
            gr, gc = divmod(goal.index(val), 3)
            dist += abs(r - gr) + abs(c - gc)
    return dist

def get_neighbors(state):
    neighbors = []
    z = state.index(0)
    r, c = divmod(z, 3)
    for dr, dc, move in [(-1, 0, 'Up'), (1, 0, 'Down'), (0, -1, 'Left'), (0, 1, 'Right')]:
        nr, nc = r + dr, c + dc
        if 0 <= nr < 3 and 0 <= nc < 3:
            nz = nr * 3 + nc
            board = list(state)
            board[z], board[nz] = board[nz], board[z]
            neighbors.append((tuple(board), move))
    return neighbors

def solve_a_star(start, goal):
    # Priority queue stores tuples: (f_score, g_score, current_state, path_taken)
    queue = [(manhattan(start, goal), 0, start, [])]
    visited = {start: 0}

    while queue:
        _, g, state, path = heapq.heappop(queue)
        if state == goal:
            return path, g
        for neighbor, move in get_neighbors(state):
            if neighbor not in visited or g + 1 < visited[neighbor]:
                visited[neighbor] = g + 1
                heapq.heappush(queue, (g + 1 + manhattan(neighbor, goal), g + 1, neighbor, path + [(move, neighbor)]))
    return None, -1

def print_board(b):
    for i in range(0, 9, 3):
        print(list(b[i:i+3]))
    print()

initial_state = (2, 8, 3, 1, 6, 4, 7, 0, 5)
goal_state = (1, 2, 3, 8, 0, 4, 7, 6, 5)

print("Initial Board:")
print_board(initial_state)
print("Target Final Board:")
print_board(goal_state)

path, steps = solve_a_star(initial_state, goal_state)
if path:
    print(f"Solved in {steps} moves!\n")
    for move, board in path:
        print(f"Move: {move}")
        print_board(board)
else:
    print("No solution found.")
```
