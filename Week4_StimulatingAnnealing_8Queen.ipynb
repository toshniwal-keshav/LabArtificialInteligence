```python
import random
import math


# Calculate number of conflicts
def cost(board):
    conflicts = 0
    n = len(board)

    for i in range(n):
        for j in range(i + 1, n):

            # Same row
            if board[i] == board[j]:
                conflicts += 1

            # Same diagonal
            elif abs(board[i] - board[j]) == abs(i - j):
                conflicts += 1

    return conflicts

# Generate a neighboring state
def get_neighbor(board):
    new_board = board.copy()

    # Pick a random column
    col = random.randint(0, 7)

    # Pick a different random row
    old_row = new_board[col]
    new_row = random.randint(0, 7)

    while new_row == old_row:
        new_row = random.randint(0, 7)

    new_board[col] = new_row

    return new_board


# Simulated Annealing
def simulated_annealing():

    # Initial state
    current = [random.randint(0, 7) for _ in range(8)]

    # Initial temperature
    T = 10.0

    # Cooling factor
    alpha = 0.90

    while True:

        current_cost = cost(current)

        # Solution found
        if current_cost == 0:
            return current

        # Generate neighboring state
        next_state = get_neighbor(current)

        next_cost = cost(next_state)

        # ΔE = New Cost - Current Cost
        delta_E = next_cost - current_cost

        # Better move
        if delta_E < 0:
            current = next_state

        # Worse move
        else:
            probability = math.exp(-delta_E / T)

            if random.random() < probability:
                current = next_state

        # Cool down
        T = T * alpha
        print("Current Temperature:", T)

    return current


# Main program
solution = simulated_annealing()

print("Final Board:", solution)
print("Final Cost:", cost(solution))
```
